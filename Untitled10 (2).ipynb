```python
# ============================================================
# CUSTOMER SEGMENTATION USING K-MEANS CLUSTERING
# ============================================================

# ============================================================
# STEP 1: IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

print("=" * 60)
print("       CUSTOMER SEGMENTATION USING K-MEANS")
print("=" * 60)


# ============================================================
# STEP 2: CREATE MALL CUSTOMER DATASET
# ============================================================
# This creates a dataset directly so the project can run
# without requiring a CSV upload.

np.random.seed(42)

n = 200

age = np.random.randint(18, 70, n)
income = np.random.randint(15, 140, n)
spending_score = np.random.randint(1, 101, n)

df = pd.DataFrame({
    "CustomerID": range(1, n + 1),
    "Age": age,
    "Annual_Income_k": income,
    "Spending_Score": spending_score
})


# ============================================================
# STEP 3: DISPLAY DATASET
# ============================================================

print("\nFirst 10 Customers:")
print(df.head(10))

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================================
# STEP 4: BASIC STATISTICS
# ============================================================

print("\n" + "=" * 60)
print("STATISTICAL SUMMARY")
print("=" * 60)

print(df.describe())


# ============================================================
# STEP 5: VISUALIZE CUSTOMER DATA
# ============================================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Annual_Income_k",
    y="Spending_Score"
)

plt.title("Annual Income vs Spending Score")
plt.xlabel("Annual Income (Thousands)")
plt.ylabel("Spending Score")

plt.show()


# ============================================================
# STEP 6: SELECT FEATURES FOR CLUSTERING
# ============================================================

# We use:
# Annual Income
# Spending Score

X = df[
    [
        "Annual_Income_k",
        "Spending_Score"
    ]
]

print("\nFeatures used for clustering:")
print(X.head())


# ============================================================
# STEP 7: STANDARDIZE FEATURES
# ============================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("\nData standardized successfully!")

print("\nFirst 5 scaled records:")
print(X_scaled[:5])


# ============================================================
# STEP 8: ELBOW METHOD
# ============================================================

inertia = []

K = range(1, 11)

for k in K:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    inertia.append(
        kmeans.inertia_
    )


# Plot Elbow Curve

plt.figure(figsize=(8, 5))

plt.plot(
    K,
    inertia,
    marker="o"
)

plt.title("Elbow Method for Optimal K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.xticks(K)

plt.show()


# ============================================================
# STEP 9: FIND SILHOUETTE SCORE
# ============================================================

silhouette_scores = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(
        X_scaled
    )

    score = silhouette_score(
        X_scaled,
        labels
    )

    silhouette_scores.append(
        score
    )


print("\nSilhouette Scores:")

for k, score in zip(
    range(2, 11),
    silhouette_scores
):

    print(
        "K =", k,
        "Score =", round(score, 3)
    )


# Plot Silhouette Scores

plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    silhouette_scores,
    marker="o"
)

plt.title("Silhouette Score")
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")

plt.show()


# ============================================================
# STEP 10: CHOOSE NUMBER OF CLUSTERS
# ============================================================
# For this project we use 5 clusters,
# similar to the traditional Mall Customers project.

optimal_k = 5

print("\nSelected number of clusters:", optimal_k)


# ============================================================
# STEP 11: TRAIN K-MEANS MODEL
# ============================================================

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(
    X_scaled
)

print("\nK-Means clustering completed!")


# ============================================================
# STEP 12: DISPLAY CLUSTERED DATA
# ============================================================

print("\nClustered Customer Data:")

print(
    df.head(20)
)


# ============================================================
# STEP 13: COUNT CUSTOMERS IN EACH CLUSTER
# ============================================================

cluster_counts = df["Cluster"].value_counts().sort_index()

print("\nCustomers in each cluster:")

print(cluster_counts)


# ============================================================
# STEP 14: VISUALIZE CLUSTERS
# ============================================================

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=df,
    x="Annual_Income_k",
    y="Spending_Score",
    hue="Cluster",
    palette="viridis",
    s=100
)

# Plot cluster centers

centers_original = scaler.inverse_transform(
    kmeans.cluster_centers_
)

plt.scatter(
    centers_original[:, 0],
    centers_original[:, 1],
    marker="X",
    s=300,
    c="red",
    label="Cluster Centers"
)

plt.title("Customer Segmentation using K-Means")
plt.xlabel("Annual Income (Thousands)")
plt.ylabel("Spending Score")

plt.legend()

plt.show()


# ============================================================
# STEP 15: CLUSTER CENTERS
# ============================================================

centers = pd.DataFrame(
    centers_original,
    columns=[
        "Annual_Income_k",
        "Spending_Score"
    ]
)

print("\nCluster Centers:")

print(
    centers
)


# ============================================================
# STEP 16: ANALYZE CUSTOMER SEGMENTS
# ============================================================

cluster_summary = df.groupby(
    "Cluster"
).agg({
    "Age": "mean",
    "Annual_Income_k": "mean",
    "Spending_Score": "mean",
    "CustomerID": "count"
}).round(2)

cluster_summary.rename(
    columns={
        "Age": "Average Age",
        "Annual_Income_k": "Average Income",
        "Spending_Score": "Average Spending Score",
        "CustomerID": "Customer Count"
    },
    inplace=True
)

print("\n" + "=" * 60)
print("CUSTOMER SEGMENT SUMMARY")
print("=" * 60)

print(
    cluster_summary
)


# ============================================================
# STEP 17: GIVE MEANINGFUL NAMES TO CLUSTERS
# ============================================================

def segment_name(row):

    income = row["Average Income"]
    spending = row["Average Spending Score"]

    if income >= 70 and spending >= 60:
        return "High Income - High Spending"

    elif income >= 70 and spending < 60:
        return "High Income - Low Spending"

    elif income < 70 and spending >= 60:
        return "Low Income - High Spending"

    else:
        return "Low Income - Low Spending"


cluster_summary["Segment"] = cluster_summary.apply(
    segment_name,
    axis=1
)

print("\nCustomer Segment Names:")

print(
    cluster_summary
)


# ============================================================
# STEP 18: CUSTOMER SEGMENT BAR CHART
# ============================================================

plt.figure(figsize=(10, 6))

cluster_summary[
    "Customer Count"
].plot(
    kind="bar"
)

plt.title("Customers per Cluster")
plt.xlabel("Cluster")
plt.ylabel("Number of Customers")

plt.xticks(rotation=0)

plt.show()


# ============================================================
# STEP 19: 3D VISUALIZATION
# ============================================================
# Uses Age, Income and Spending Score.

fig = plt.figure(figsize=(10, 7))

ax = fig.add_subplot(
    111,
    projection="3d"
)

scatter = ax.scatter(
    df["Age"],
    df["Annual_Income_k"],
    df["Spending_Score"],
    c=df["Cluster"],
    cmap="viridis",
    s=60
)

ax.set_xlabel("Age")
ax.set_ylabel("Annual Income")
ax.set_zlabel("Spending Score")

ax.set_title(
    "3D Customer Segmentation"
)

plt.show()


# ============================================================
# STEP 20: PREDICT NEW CUSTOMER SEGMENT
# ============================================================

print("\n" + "=" * 60)
print("NEW CUSTOMER PREDICTION")
print("=" * 60)


# Example customer

new_customer = pd.DataFrame({
    "Annual_Income_k": [90],
    "Spending_Score": [85]
})


# Scale new customer

new_customer_scaled = scaler.transform(
    new_customer
)


# Predict cluster

new_cluster = kmeans.predict(
    new_customer_scaled
)


print(
    "\nPredicted Cluster:",
    new_cluster[0]
)


# Get segment name

predicted_segment = cluster_summary.loc[
    new_cluster[0],
    "Segment"
]

print(
    "Predicted Customer Segment:",
    predicted_segment
)


# ============================================================
# STEP 21: USER INPUT CUSTOMER
# ============================================================

print("\n" + "=" * 60)
print("ENTER NEW CUSTOMER DETAILS")
print("=" * 60)

user_income = float(
    input(
        "Annual Income (in thousands): "
    )
)

user_spending = float(
    input(
        "Spending Score (1-100): "
    )
)


# Create input DataFrame

user_customer = pd.DataFrame({
    "Annual_Income_k": [
        user_income
    ],
    "Spending_Score": [
        user_spending
    ]
})


# Scale input

user_scaled = scaler.transform(
    user_customer
)


# Predict

user_cluster = kmeans.predict(
    user_scaled
)[0]


print("\n" + "=" * 60)
print("CUSTOMER SEGMENT RESULT")
print("=" * 60)

print(
    "Cluster:",
    user_cluster
)

print(
    "Customer Segment:",
    cluster_summary.loc[
        user_cluster,
        "Segment"
    ]
)


# ============================================================
# STEP 22: VISUALIZE NEW CUSTOMER
# ============================================================

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=df,
    x="Annual_Income_k",
    y="Spending_Score",
    hue="Cluster",
    palette="viridis",
    s=80
)

plt.scatter(
    user_income,
    user_spending,
    marker="*",
    s=400,
    c="red",
    label="New Customer"
)

plt.title(
    "New Customer on Segmentation Map"
)

plt.xlabel(
    "Annual Income (Thousands)"
)

plt.ylabel(
    "Spending Score"
)

plt.legend()

plt.show()


# ============================================================
# STEP 23: SAVE MODEL
# ============================================================

joblib.dump(
    kmeans,
    "customer_segmentation_model.pkl"
)

joblib.dump(
    scaler,
    "customer_scaler.pkl"
)

print("\nModels saved successfully!")

print(
    "customer_segmentation_model.pkl"
)

print(
    "customer_scaler.pkl"
)


# ============================================================
# STEP 24: LOAD MODEL
# ============================================================

loaded_kmeans = joblib.load(
    "customer_segmentation_model.pkl"
)

loaded_scaler = joblib.load(
    "customer_scaler.pkl"
)

print("\nModels loaded successfully!")


# ============================================================
# STEP 25: FINAL MESSAGE
# ============================================================

print("\n" + "=" * 60)
print("       PROJECT COMPLETED SUCCESSFULLY!")
print("=" * 60)

print("""
Project:
Customer Segmentation using K-Means Clustering

Algorithm:
K-Means Clustering

Features:
- Annual Income
- Spending Score

Techniques:
- Data preprocessing
- Feature scaling
- Elbow Method
- Silhouette Score
- K-Means clustering
- 2D visualization
- 3D visualization
- New customer prediction
- Model saving
""")
```
